**Drop Null values**

```python
data  =data.dropna()
data
# Drop Null values
```

**Top 10 best selling brand**

```python
print(data['Brand'].value_counts().head()) #Count the top five brands
data['Brand'].value_counts().head().plot(kind = 'bar') #Plot the top five brands as a bar chart
```

**Premia is the leading brand with the highest count of 155, followed by Faces Canada at 99, Insight at 82, Amul at 79, and Lakme at 74. The bar chart below clearly illustrates the dominance of these top brands.**

**top 10 most selling category**

```python
print(data['Category'].value_counts().head(6)) #Count and print the top six categories
data['Category'].value_counts().head(6).plot(kind = 'pie', autopct = '%1.1f%%')
plt.show()
```


**Personal Care is the most frequent category, making up a significant proportion of the dataset at 1224 entries.Packaged Food and Home & Kitchen follow, with 1117 and 778 entries respectively, highlighting their importance.Dairy & Beverages and Specials are less common, with 426 and 163 entries, respectively, showing their relative rarity compared to the other categories.**
